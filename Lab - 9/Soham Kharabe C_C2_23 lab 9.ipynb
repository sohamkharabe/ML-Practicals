```python
import warnings
warnings.filterwarnings("ignore")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import (silhouette_score)
from sklearn.decomposition import PCA

from scipy.cluster.hierarchy import dendrogram, linkage
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
df = pd.read_csv("w.csv")
```

```python
print(df.shape)
print(df.head())
print(df.columns.tolist())
print(df.info())
```

```python
print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

# Remove duplicate records if any
df = df.drop_duplicates().copy()

# Fill missing numeric values with the median
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())

print("\nMissing Values After Treatment:")
print(df.isnull().sum())

print("\nDataset Shape After Cleaning:", df.shape)
```

```python
features = [
    'Fresh', 'Milk', 'Grocery',
    'Frozen', 'Detergents_Paper', 'Delicassen'
]

# Histograms
df[features].hist(figsize=(12, 8), bins=20)
plt.suptitle("Distribution of Customer Spending")
plt.tight_layout()
plt.show()

# Correlation heatmap
plt.figure(figsize=(10, 7))
sns.heatmap(
    df[features].corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)
plt.title("Correlation Between Spending Categories")
plt.show()

# Boxplots for detecting potential outliers
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[features])
plt.title("Outlier Analysis of Spending Categories")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
X = df[features].copy()

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("Scaled Data Shape:", X_scaled.shape)
print("\nFirst 5 Scaled Records:")
print(pd.DataFrame(
    X_scaled,
    columns=features
).head().round(2))
```

```python
wcss = []

for k in range(1, 11):
    model = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    model.fit(X_scaled)
    wcss.append(model.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o')
plt.xlabel("Number of Clusters (K)")
plt.ylabel("Within-Cluster Sum of Squares (WCSS)")
plt.title("Elbow Method for Optimal K")
plt.xticks(range(1, 11))
plt.grid(True)
plt.show()
```

```python
## kmeans = KMeans(
    n_clusters=3,
    random_state=42,
    n_init=10
)

df['KMeans_Cluster'] = kmeans.fit_predict(X_scaled)

print("K-Means Cluster Counts:")
print(df['KMeans_Cluster'].value_counts().sort_index())

print("\nAverage Spending by K-Means Cluster:")
print(
    df.groupby('KMeans_Cluster')[features]
      .mean()
      .round(2)
)
```

```python
hierarchical = AgglomerativeClustering(
    n_clusters=3,
    linkage='ward'
)

df['Hierarchical_Cluster'] = hierarchical.fit_predict(X_scaled)

print("Hierarchical Cluster Counts:")
print(df['Hierarchical_Cluster'].value_counts().sort_index())

print("\nAverage Spending by Hierarchical Cluster:")
print(
    df.groupby('Hierarchical_Cluster')[features]
      .mean()
      .round(2)
)
```

```python
linked = linkage(X_scaled, method='ward')

plt.figure(figsize=(14, 6))
dendrogram(
    linked,
    truncate_mode='lastp',
    p=30,
    leaf_rotation=90,
    leaf_font_size=9
)

plt.title("Hierarchical Clustering Dendrogram")
plt.xlabel("Customer Groups")
plt.ylabel("Ward Linkage Distance")
plt.tight_layout()
plt.show()
```

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

plot_df = pd.DataFrame({
    'PC1': X_pca[:, 0],
    'PC2': X_pca[:, 1],
    'KMeans_Cluster': df['KMeans_Cluster'],
    'Hierarchical_Cluster': df['Hierarchical_Cluster']
})

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(
    data=plot_df,
    x='PC1', y='PC2',
    hue='KMeans_Cluster',
    palette='Set1',
    ax=axes[0]
)
axes[0].set_title("K-Means Customer Segments")

sns.scatterplot(
    data=plot_df,
    x='PC1', y='PC2',
    hue='Hierarchical_Cluster',
    palette='Set2',
    ax=axes[1]
)
axes[1].set_title("Hierarchical Customer Segments")

plt.tight_layout()
plt.show()

print(
    "Variance explained by first 2 PCs:",
    round(pca.explained_variance_ratio_.sum() * 100, 2),
    "%"
)
```

```python
from sklearn.metrics import silhouette_score, davies_bouldin_score
comparison = pd.DataFrame({
    'Algorithm': ['K-Means', 'Hierarchical'],
    'Silhouette Score': [
        silhouette_score(X_scaled, df['KMeans_Cluster']),
        silhouette_score(X_scaled, df['Hierarchical_Cluster'])
    ],
    'Davies-Bouldin Index': [
        davies_bouldin_score(X_scaled, df['KMeans_Cluster']),
        davies_bouldin_score(X_scaled, df['Hierarchical_Cluster'])
    ]
})

print(comparison.round(3))

comparison.set_index('Algorithm').plot(
    kind='bar',
    figsize=(9, 5),
    subplots=True,
    layout=(1, 2),
    legend=False,
    rot=0
)

plt.tight_layout()
plt.show()
```
