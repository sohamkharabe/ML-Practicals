```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
```

```python
import numpy as np
def find_s(X,y):
    hypothesis = None
    print("Initial Hypothesis :",hypothesis)
    for i in range(len(X)):
        if(y[i] == "Yes"):
            if hypothesis is None:
                hypothesis = X[i].copy()
            else:
                for j in range(len(hypothesis)):
                    if hypothesis[j] != X[i][j]:
                        hypothesis[j] = "?"
        print(f"\nAfter Training Example {i+1}")
        print(hypothesis)
    return hypothesis
final_hypothesis = find_s(X,y)
```

### Candidate Elimination Algorithm

```python
import copy

def candidate_elimination(concepts, target):
    specefic = concepts[0].copy()
    general = [["?" for i in range(len(specefic))]]

    print("\nInitial S:", specefic)
    print("\nInitial G:", general)

    for i, h in enumerate(concepts):
        if target[i] == "Yes":
            for x in range(len(specefic)):
                if h[x] != specefic[x]:
                    specefic[x] = "?"
        else:
            for x in range(len(specefic)):
                if h[x] != specefic[x]:
                    general.append(
                        ["?" if j != x else specefic[x] for j in range(len(specefic))]
                    )

        print("\nExample", i + 1)
        print("S =", specefic)
        print("G =", general)

    return specefic, general


S, G = candidate_elimination(X, y)

print("\nFinal Specefic Boundary")
print(S)

print("\nFinal General Boundary")
for g in G:
    print(g)
```
