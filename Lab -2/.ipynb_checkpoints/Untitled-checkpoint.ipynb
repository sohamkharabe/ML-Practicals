### Simple Linear Regression

```python
X = df[['medianIncome']]
y = df['medianHouseValue']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 1300
)
```

```python
slr = LinearRegression()
slr.fit(X_train,y_train)
```

```python
y_pred = slr.predict(X_test)
```

```python
plt.figure(figsize=(8,5))
plt.scatter(X_test,y_test,color="blue")
plt.plot(X_test,y_pred,color="red")
plt.xlabel("Median Income")
plt.ylabel("Household Value")
plt.show()
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)
print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2-Score: ",r2)
```

### Multiple Linear Regression

```python
df.columns
```

```python
X = df.drop("medianHouseValue",axis=1)
y = df["medianHouseValue"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 1300
)
```

```python
mlr = LinearRegression()
mlr.fit(X_train,y_train)
```

```python
y_pred = mlr.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)
print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2-Score: ",r2)
```

```python
plt.figure(figsize=(7,6))
plt.scatter(y_test,y_pred,color="blue")

plt.plot([y_test.min(),y_test.max()],
         [y_test.min(),y_test.max()],
         color="red",linewidth=2)

plt.xlabel("Actual Household Value")
plt.ylabel("Predicted Household Value")
plt.title("Actual v/s Predicted Values")
plt.show()
```

## Hyperparameters Tunning

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error , mean_squared_error , r2_score
import numpy as np
```

```python
lr = LinearRegression()
```

```python
lr_params = {
    'fit_intercept' : [True,False],
    'positive' : [True,False]
}
```

```python
grid_search = GridSearchCV(
    estimator = lr,
    param_grid = lr_params,
    cv = 5,
    scoring = 'r2',
    n_jobs = -1
)
```

```python
grid_search.fit(X_train, y_train)
```

```python
print("Best Parameters :")
print(grid_search.best_score_)
```

```python
best_model = grid_search.best_estimator_
```

```python
LinearRegression(
    fit_intercept = True,
    positive = False
)
```

```python
y_pred = best_model.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,y_pred)
mse = mean_squared_error(y_test,y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,y_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2-Score: ",r2)
```

```python
print(best_model)
```

```python
results = grid_search.cv_results_

for mean_score, params in zip(results['mean_test_score'],results['params']):
    print(params, "R² = ", round(mean_score,4))
```

## Ridge Regression

```python
from sklearn.model_selection import GridSearchCV
```

```python
from sklearn.linear_model import Ridge

ridge = Ridge()
ridge.fit(X_train, y_train)
ridge_pred = ridge.predict(X_test)
print("Default Ridge R² : ", r2_score(y_test,ridge_pred))
```

```python
param_grid = {'alpha' : [0.001,0.01,0.1,1,10,100]}
```

```python
grid_ridge = GridSearchCV(estimator=Ridge(),param_grid = param_grid,scoring = 'r2',cv = 5)
```

```python
grid_ridge.fit(X_train,y_train)
```

```python
print("Best Alpha: ", grid_ridge.best_params_)
```

```python
print(grid_ridge.best_score_)
```

```python
best_ridge = grid_ridge.best_estimator_
ridge_pred = best_ridge.predict(X_test)
```

```python
mae = mean_absolute_error(y_test,ridge_pred)
mse = mean_squared_error(y_test,ridge_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test,ridge_pred)

print("MAE: ",mae)
print("MSE: ",mse)
print("RMSE: ",rmse)
print("R2-Score: ",r2)
```

## Lasso Regression

```python
from sklearn.linear_model import Lasso

param_grid = {
    'alpha':[0.001,0.01,0.1,1,10]}
```

```python
grid_lasso = GridSearchCV(
    Lasso(max_iter = 5000),
    param_grid,cv = 5,
    scoring = 'r2'
)
grid_lasso.fit(X_train,y_train)
```

```python
print(grid_lasso.best_params_)
```

```python
lasso_pred = grid_lasso.predict(X_test)
print(r2_score(y_test,lasso_pred))
```
