```python
Q1 = df["Fare"].quantile(0.25)
Q3 = df["Fare"].quantile(0.75)

IQR = Q3-Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df["Fare"] >= lower) & (df["Fare"] <= upper)]
```

```python
sns.boxplot(x=df["Fare"])
plt.show()
```

```python
le = LabelEncoder()

df["Sex"] = le.fit_transform(df["Sex"])
```
