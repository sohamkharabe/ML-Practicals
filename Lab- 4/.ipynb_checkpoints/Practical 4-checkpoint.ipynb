## Decision Tree Classification Algorithm

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score,confusion_matrix,classification_report
from sklearn.inspection import permutation_importance
```

```python
X = df.drop('quality',axis = 1)
y=df['quality']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size = 0.20,
    random_state = 42,
    stratify=y
)
results=[]
```

```python
def evaluate(name,model):
    model.fit(X_train,y_train)
    pred=model.predict(X_test)
    acc=accuracy_score(y_test,pred)
    print(name,acc)
    print(classification_report(y_test,pred))
    cm = confusion_matrix(y_test,pred)
    sns.heatmap(cm,annot=True,fmt='d',cmap='Blues')
    plt.title(name)
    plt.show()
    results.append([name, acc, model.get_depth(), model.tree_.node_count])
    return model
```

```python
tree=evaluate('Original',DecisionTreeClassifier(criterion='entropy',random_state=42))
plt.figure(figsize=(35,15))
plot_tree(tree,feature_names=X.columns,filled=True,max_depth=3)
plt.show()
```

```python
plt.figure(figsize=(25,15))
plot_tree(
    tree,
    feature_names=X.columns,
    class_names=[str(i) for i in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=12,
    max_depth=3
)
plt.show()
```

```python
plt.figure(figsize=(25,15))
plot_tree(
    tree,
    feature_names=X.columns,
    class_names=[str(i) for i in sorted(y.unique())],
    filled=True,
    rounded=True,
    fontsize=12,
    max_depth=20
)
plt.show()
```
